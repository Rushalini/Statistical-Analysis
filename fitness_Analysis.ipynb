```python
import statistics
import matplotlib.pyplot as plt
m=["raju","ravi","shalini","reena","joy"]
age=[23,22,20,18,17]
workout_hours=[2.5,3,1.5,2,1.5]
cal_burns=[300,1500,500,300,480]
print("--------Fitness Dataset--------")
print("\n")
for i in range(len(m)):
    print(m[i],age[i],workout_hours[i],cal_burns[i])
print("\n")
avg_workout=sum(workout_hours)/len(workout_hours)
print("Average workout Hours=",avg_workout)
highest_cal_burns=max(cal_burns)
print("Highest calories burned=",highest_cal_burns)
lowest_workout=min(workout_hours)
print("Lowest workout hours=",lowest_workout)
median_cal_burns=statistics.median(cal_burns)
print("Median calories burned=",median_cal_burns)
std_dev=statistics.stdev(cal_burns)
print("Standard Deviation=",std_dev)
index=cal_burns.index(highest_cal_burns)
print("Person with highest calories burned=",m[index])
plt.bar(m,cal_burns)
plt.xlabel("Member Names")
plt.ylabel("Calories Burned")
plt.title("Fitness Analysis")
plt.show()
```
